5. Write a program to develop a decision tree classifier to predict signal quality (Good or
Bad) based on transmitter, signal strength, frequency.

Transmitter Signal Strength Frequency Signal Quality

T1 75 2000 Good

T2 50 1500 Bad

T1 90 2200 Good

T3 40 1300 Bad

T2 85 2100 Good

T1 70 1800 Good

T3 30 1200 Bad

T2 60 1600 Bad

T1 80 1900 Good

T3 20 1100 Bad

```python
import pandas as pd

# Data for signal quality prediction
signal_quality_data = {
    'Transmitter': ['T1', 'T2', 'T1', 'T3', 'T2', 'T1', 'T3', 'T2', 'T1', 'T3'],
    'Signal Strength': [75, 50, 90, 40, 85, 70, 30, 60, 80, 20],
    'Frequency': [2000, 1500, 2200, 1300, 2100, 1800, 1200, 1600, 1900, 1100],
    'Signal Quality': ['Good', 'Bad', 'Good', 'Bad', 'Good', 'Good', 'Bad', 'Bad', 'Good', 'Bad']
}

df_signal_quality = pd.DataFrame(signal_quality_data)
display(df_signal_quality.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# One-hot encode the 'Transmitter' column
df_signal_quality_encoded = pd.get_dummies(df_signal_quality, columns=['Transmitter'], drop_first=True)

# Label encode the 'Signal Quality' target variable
label_encoder = LabelEncoder()
df_signal_quality_encoded['Signal Quality_encoded'] = label_encoder.fit_transform(df_signal_quality_encoded['Signal Quality'])

# Define features (X_signal) and target (y_signal)
X_signal = df_signal_quality_encoded.drop(columns=['Signal Quality', 'Signal Quality_encoded'])
y_signal = df_signal_quality_encoded['Signal Quality_encoded']

print("Encoded DataFrame head:")
display(df_signal_quality_encoded.head())

print("Features (X_signal) shape:", X_signal.shape)
print("Target (y_signal) shape:", y_signal.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Split the data into training and testing sets
X_signal_train, X_signal_test, y_signal_train, y_signal_test = train_test_split(X_signal, y_signal, test_size=0.3, random_state=42)

# Initialize and train the Decision Tree Classifier
dtc = DecisionTreeClassifier(random_state=42)
dtc.fit(X_signal_train, y_signal_train)

# Make predictions on the test data
y_signal_pred = dtc.predict(X_signal_test)

# Evaluate the model
print("Decision Tree Classifier Performance for Signal Quality Prediction:")
print(f"Accuracy: {accuracy_score(y_signal_test, y_signal_pred):.2f}")
print("\nClassification Report:")
print(classification_report(y_signal_test, y_signal_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_signal_test, y_signal_pred))
```
