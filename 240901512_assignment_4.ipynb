4. Write a program to build a machine learning linear regression model to predict
the price of a house based on its features. You are given a dataset containing
information of Year and price.

Year Price(₹Lakhs)

2010 25

2011 28

2012 30

2013 34

2014 37

2015 40

2016 43

2017 45

2018 47

2019 50

```python
import pandas as pd

# Data for house price prediction
house_data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019],
    'Price (₹ Lakhs)': [25, 28, 30, 34, 37, 40, 43, 45, 47, 50]
}

df_house = pd.DataFrame(house_data)
display(df_house.head())
```

```python
# Define features (X_house) and target (y_house)
X_house = df_house[['Year']]
y_house = df_house['Price (₹ Lakhs)']

print("Features (X_house) shape:", X_house.shape)
print("Target (y_house) shape:", y_house.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split the house price data into training and testing sets
X_house_train, X_house_test, y_house_train, y_house_test = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

print("X_house_train shape:", X_house_train.shape)
print("X_house_test shape:", X_house_test.shape)
print("y_house_train shape:", y_house_train.shape)
print("y_house_test shape:", y_house_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model for house price prediction
linear_reg_house_model = LinearRegression()

# Train the model
linear_reg_house_model.fit(X_house_train, y_house_train)

# Make predictions on the test data
y_house_pred = linear_reg_house_model.predict(X_house_test)

print("Predictions on house price test data:\n", y_house_pred)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Evaluate the Linear Regression model for house price prediction
mae_house = mean_absolute_error(y_house_test, y_house_pred)
mse_house = mean_squared_error(y_house_test, y_house_pred)
rmse_house = np.sqrt(mse_house)
r2_house = r2_score(y_house_test, y_house_pred)

print(f"House Price Prediction - Mean Absolute Error (MAE): {mae_house:.2f}")
print(f"House Price Prediction - Mean Squared Error (MSE): {mse_house:.2f}")
print(f"House Price Prediction - Root Mean Squared Error (RMSE): {rmse_house:.2f}")
print(f"House Price Prediction - R-squared (R2): {r2_house:.2f}")
```
